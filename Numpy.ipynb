```python

a = np.array([1, 2, 3, 4, 5])           # from list
print("this is a:", a)
b = np.zeros((3, 4))                    # 3x4 zeros
print("this is b:", b)
c = np.ones((2, 3))                     # 2x3 ones
print("this is c:", c)
d = np.full((2, 2), 7)                  # filled with 7
print("this is d:", d)
e = np.arange(0, 10, 2)                 # [0 2 4 6 8]
print("this is e:", e)
f = np.linspace(0, 1, 5)                # 5 evenly spaced values 0→1
print("this is f:", f)

print("Shape of a:", a.shape)
print("Data type of b:", b.dtype)
print("Shape of f:", f.shape)
```

```python
img = np.random.rand(28, 28)
flat = img.reshape(784)        # flatten for ML model
print("Shape of flat:", flat.shape)   # (784,)
back = flat.reshape(28, 28)    # restore
print("Shape of back:", back.shape)   # (28, 28)
assert np.allclose(img, back)

# -1 means "auto-compute this dimension"
m = np.arange(12).reshape(3, -1)   # (3, 4)
print("Shape of m:", m.shape)   # (3, 4)
```

```python
data = np.arange(20).reshape(4, 5)
print("data:", data)
print("data[1, 2]:", data[1, 2])       # row 1, col 2 -> 7
print("data[0]:", data[0])          # first row
print("data[:, 1:3]:", data[:, 1:3])     # all rows, cols 1-2
print("data[::2]:", data[::2])        # every other row
print("data[::-1]:", data[::-1])       # reverse rows
```

```python
a = np.arange(10)
v = a[2:5]              # VIEW: shares memory
v[0] = 99
print("a:", a)                # [0 1 99 3 4 ...] -> original changed!

c = a.copy()            # COPY: independent
c[0] = -1
print("a[0]:", a[0])             # still 0
```

```python
scores = np.array([45, 78, 92, 60, 88])
print("Scores:", scores)
passed = np.where(scores >= 60, 1, 0)     # [0 1 1 1 1]
print("Passed:", passed)
''' how does where condition works: np.where(condition, value_if_true, value_if_false) '''

# Clip outliers
clean = np.clip(scores, 50, 85)
print("Cleaned scores:", clean)
''' how does clip function works: np.clip(array, min_value, max_value)
 - values below min_value are set to min_value and values above max_value are set to max_value '''
```

```python
X = np.c_[np.ones(100), np.random.rand(100)]   # add intercept column
y_true_w = np.array([3.0, 5.0])
y = X @ y_true_w + np.random.randn(100) * 0.1

w_hat = np.linalg.inv(X.T @ X) @ X.T @ y       # recovered ≈ [3, 5]
print(w_hat.round(2))
```

```python
X = np.random.randn(200, 2)
X[:, 1] = X[:, 0] * 0.8 + X[:, 1] * 0.2   # correlated features
cov = np.cov(X.T)
vals, vecs = np.linalg.eigh(cov)
principal_axis = vecs[:, np.argmax(vals)]  # direction of max variance
print(principal_axis)
```

```python
rng = np.random.default_rng(7)
data = np.vstack([
    rng.normal([2, 2], 0.6, (150, 2)),
    rng.normal([8, 8], 0.6, (150, 2)),
    rng.normal([8, 2], 0.6, (150, 2)),
])

centroids = data[rng.choice(len(data), 3, replace=False)]
for _ in range(10):
    # assign each point to nearest centroid
    d = ((data[:, None, :] - centroids)**2).sum(-1)   # (n, k) distances
    labels = d.argmin(axis=1)
    # recompute centroids
    centroids = np.array([data[labels == k].mean(axis=0) for k in range(3)])

print(centroids.round(1))   # ≈ [[2,2],[8,2],[8,8]]
```
