```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

Load Your Excel File

```python
df = pd.read_excel("students.xlsx")
df
```

Bar Plot (Marks Comparison for Each Student)

```python
df.plot(x="Name", y=["Math", "Science", "English"], kind="bar")
plt.title("Marks Comparison of Students")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.xticks(rotation=0)
plt.show()
```

Scatter Plot (Math vs Science)

```python
sns.scatterplot(x="Math", y="Science", hue="Name", data=df,marker='s')
plt.title("Math vs Science Marks")
plt.show()
```

Scatter Plot (Math vs English)

```python
sns.scatterplot(x="Math", y="English", hue="Name", data=df,marker='*',s=120)
plt.title("Math vs English Marks")
plt.show()

```

Line Plot (Subject-wise performance)

```python
sns.lineplot(x="Name", y="Math", data=df, marker="o", label="Math")
sns.lineplot(x="Name", y="Science", data=df, marker="o", label="Science")
sns.lineplot(x="Name", y="English", data=df, marker="o", label="English")
plt.title("Subject-wise Student Performance")
plt.ylabel("Marks")
plt.xticks(rotation=45)
plt.legend()
plt.show()

```

Box Plot (Subject Distribution)

```python
sns.boxplot(data=df[["Math", "Science", "English"]])
plt.title("Distribution of Marks in Each Subject")
plt.show()
```

Violin Plot (Subject Distribution)

```python
sns.violinplot(data=df[["Math", "Science", "English"]])
plt.title("Violin Plot of Subjects")
plt.show()
```

Heatmap (Correlation Between Subjects)

```python
corr = df[["Math", "Science", "English"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Between Subjects")
plt.show()
```

Pair Plot (All subjects relation)

```python
sns.pairplot(df[["Math", "Science", "English"]])
plt.show()
```

Histogram for Each Subject
Math

```python
sns.histplot(df["Math"], kde=True)
plt.title("Math Marks Distribution")
plt.show()
```

Science

```python
sns.histplot(df["Science"], kde=True)
plt.title("Science Marks Distribution")
plt.show()
```

English

```python
sns.histplot(df["English"], kde=True)
plt.title("English Marks Distribution")
plt.show()
```

Count Plot for Names

```python
sns.countplot(x="Name", data=df)
plt.title("Count of Students")
plt.xticks(rotation=45)
plt.show()
```

Average Marks Bar Plot

```python
df["Average"] = (df["Math"] + df["Science"] + df["English"]) / 3
sns.barplot(x="Name", y="Average", data=df)
plt.title("Average Marks of Each Student")
plt.xticks(rotation=45)
plt.show()
```
