```python
import matplotlib.pyplot as plt 
import numpy as np
```

🔹 1. Line Graph

```python
x = [1,20,3,4,5]
y = [2,1,6,4,10]

plt.plot(x, y)
plt.title("Line graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

🔹 2. Multiple Line Graph

```python
y1 = [2,6,5,1,6]

y2 = [1,5,3,5,6]
y3 = [1,4,2,4,6]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")
plt.plot(x, y3, label="Line 3")
plt.title("Multiple Lines")
plt.legend()
plt.show()

```

```python
x = [2010, 2011, 2012, 2013,2014,2015,2015]
y = [5, 7, 3, 8,5,7,3]

plt.bar(x, y)
plt.title("Bar Chart")
plt.show()
```

🔹 4. Horizontal Bar Chart

```python
x = [2010, 2011, 2012, 2013,2014,2015,2015]
y = [5, 7, 10, 8,5,7,3]

plt.barh(x, y)
plt.title("Horizontal Bar Chart")
plt.show()
```

🔹 5. Stacked Bar Chart

```python
boys = [11,18,20,60]
girls = [14,16,18,20]
x = np.arange(4)

plt.bar(x, boys, label="Boys")
plt.bar(x, girls, bottom=boys, label="Girls")
plt.xticks(x, ["Class 1","Class 2","Class 3","Class 4"])
plt.title("Stacked Bar Chart")
plt.legend()
plt.show()
```

🔹 6. Scatter Plot

```python
p = [5,7,8,7,8,17,7,9,4,11]
q = [99,86,87,88,100,86,103,87,94,78]

plt.scatter(p, q)
plt.title("Scatter Plot")
plt.show()
```

🔹 7. Pie Chart

```python
sizes = [20, 30, 25]
labels = ["A", "B", "C"]

plt.pie(sizes, labels=labels, autopct="%1.2f%%")
plt.title("Pie Chart")
plt.show()
```

🔹 8. Histogram

```python
data = np.random.randn(100)

plt.hist(data, bins=20, color="skyblue", edgecolor="black")
plt.title("Histogram")
plt.show()
```

🔹 9. Subplots (Many Plots Together)

```python
x = np.linspace(0, 10, 100)

plt.subplot(2,2,1)
plt.plot(x, np.sin(x))
plt.title("Sine")

plt.subplot(2,2,2)
plt.plot(x, np.cos(x))
plt.title("Cosine")

plt.subplot(2,2,3)
plt.plot(x, np.tan(x))
plt.title("Tangent")

plt.subplot(2,2,4)
plt.plot(x, np.exp(x/3))
plt.title("Exponential")

plt.tight_layout()
plt.show()
```
